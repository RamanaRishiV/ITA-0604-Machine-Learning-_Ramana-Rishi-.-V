```python

import csv
import io

# Training data stored as CSV
csv_data = """Sky,Temperature,Humidity,Wind,Water,Forecast,EnjoySport
Sunny,Warm,Normal,Strong,Warm,Same,Yes
Sunny,Warm,High,Strong,Warm,Same,Yes
Rainy,Cold,High,Strong,Warm,Change,No
Sunny,Warm,High,Strong,Cool,Change,Yes
"""

# Read CSV data
data = list(csv.reader(io.StringIO(csv_data)))
attributes = data[0][:-1]
examples = data[1:]

# Initialize Specific and General boundaries
S = ["Ø"] * len(attributes)
G = [["?"] * len(attributes)]

# Candidate-Elimination algorithm
for example in examples:
    x = example[:-1]
    target = example[-1]

    if target == "Yes":
        # Remove inconsistent hypotheses from G
        G = [g for g in G if all(
            g[i] == "?" or g[i] == x[i]
            for i in range(len(x))
        )]

        # Generalize S
        for i in range(len(x)):
            if S[i] == "Ø":
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = "?"

    else:
        # Specialize G for negative examples
        new_G = []

        for g in G:
            if all(g[i] == "?" or g[i] == x[i]
                   for i in range(len(x))):
                for i in range(len(x)):
                    if S[i] not in ("Ø", "?") and x[i] != S[i]:
                        h = g.copy()
                        h[i] = S[i]
                        new_G.append(h)
            else:
                new_G.append(g)

        G = new_G

# Display results
print("Attributes:", attributes)
print("\nFinal Specific Boundary (S):")
print(S)

print("\nFinal General Boundary (G):")
for hypothesis in G:
    print(hypothesis)
```
